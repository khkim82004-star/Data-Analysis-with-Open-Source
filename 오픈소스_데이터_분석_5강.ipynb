## 5-2 JSON 형식 저장

```python
###--- 라이브러리 임포트
import json
import pandas as pd

data = {
    "이름": "홍길동",
    "나이": 25,
    "거주지": "서울",
    "관심사": ["프로그래밍", "데이터 분석", "여행"]
}

##--- json.dump를 이용한 저장 w:기록모드
with open('output.json', mode='w', encoding='utf-8') as f:
    json.dump(data, f, indent=4, ensure_ascii=False)
    #json의 dump함수이용해서 파일로 저장 가능, 들여쓰기,

##--- DataFrame을 이용한 저장
df = pd.DataFrame(data)
df.to_json('output_df.json', orient='records', indent=4, force_ascii=False)
# orient='columns'로 하면 데이터가 보이는 게 달라짐
```
